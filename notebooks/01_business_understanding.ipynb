# 01. Entendimiento del negocio: análisis de riesgo crediticio

## Qué quiero resolver

En este proyecto actúo como analista de datos para una entidad financiera.
Mi objetivo es **entender patrones históricos asociados a dificultades de pago**
en solicitudes de crédito. No voy a construir un modelo predictivo ni a decidir
si una persona debe recibir un préstamo.

La pregunta principal que guiará mi análisis es:

> ¿Qué características presentan los segmentos con mayor tasa observada de
> dificultad de pago y cómo puedo presentar estos patrones en un dashboard
> ejecutivo de seguimiento?

Trabajo con `application_train.csv` de **Home Credit Default Risk**. La variable
`TARGET` tiene la siguiente lectura:

| Valor | Interpretación para este análisis |
|---:|---|
| `0` | Solicitud sin dificultad de pago reportada en el resultado histórico |
| `1` | Solicitud con dificultad de pago reportada |

## Qué voy a analizar

Divido mi recorrido en cuatro preguntas de negocio:

| Tema | Pregunta que voy a responder | Indicador |
|---|---|---|
| Panorama | ¿Cuántas solicitudes presentan dificultad de pago? | Total y tasa de dificultad |
| Capacidad financiera | ¿Cómo varía la tasa por ingresos, crédito y cuota? | Ratios crédito/ingreso y cuota/ingreso |
| Perfil descriptivo | ¿Qué diferencias observo por edad, educación, ocupación o vivienda? | Tasa por segmento con volumen suficiente |
| Priorización ejecutiva | ¿Qué vistas son útiles para monitorear la cartera? | KPIs y segmentos para Power BI |

### Límites que mantendré visibles

- Mi análisis describe **asociaciones históricas**, no causalidad.
- Edad, género, educación, ocupación y vivienda no deben convertirse en
  criterios aislados para negar crédito.
- La segmentación financiera es exploratoria: no es un score ni una predicción.
- Los datos fuente y el archivo limpio contienen nivel solicitante y no se
  publicarán en Git.

## Verificación inicial del insumo

Antes de comenzar quiero comprobar que el archivo requerido está en la ruta del
proyecto. Este bloque no analiza clientes; solamente valida que puedo continuar.

| Elemento del código | Qué hace | Para qué lo uso |
|---|---|---|
| `Path` | Representa rutas de carpetas y archivos | Evitar escribir rutas manuales frágiles |
| `.exists()` | Comprueba si un archivo existe | Detenerme con un mensaje claro si falta el CSV |
| `assert` | Exige que una condición sea verdadera | No ejecutar análisis sin fuente de datos |

### Lectura línea por línea del código

| Línea o instrucción | Qué estoy haciendo |
|---|---|
| `from pathlib import Path` | Importo una herramienta para trabajar con rutas de archivos de forma clara. |
| `PROJECT_ROOT = ...` | Identifico la raíz del proyecto: si estoy en la raíz uso esa carpeta; si estoy en `notebooks/`, subo un nivel. |
| `DATA_PATH = ...` | Construyo la ruta exacta donde debe estar el CSV original. |
| `assert DATA_PATH.exists(), (...)` | Detengo el notebook con un mensaje entendible si el dataset no está disponible. |
| `print(...)` | Confirmo en pantalla qué archivo voy a utilizar. |

```python
# Importo Path para construir rutas sin escribir separadores manualmente.
from pathlib import Path

# Identifico la raíz del proyecto, ejecute el notebook desde la raíz o desde notebooks/.
PROJECT_ROOT = Path.cwd() if (Path.cwd() / "data").exists() else Path.cwd().parent
# Construyo la ruta al dataset fuente que necesitaré en los siguientes notebooks.
DATA_PATH = PROJECT_ROOT / "data" / "raw" / "application_train.csv"

# Detengo la ejecución con un mensaje claro si el archivo aún no está disponible.
assert DATA_PATH.exists(), (
    "No encuentro data/raw/application_train.csv. "
    "Debo agregar el dataset antes de continuar."
)
# Confirmo en pantalla qué archivo local encontré.
print(f"Fuente disponible: {DATA_PATH.relative_to(PROJECT_ROOT)}")
```
